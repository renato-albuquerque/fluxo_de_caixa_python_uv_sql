PARTE 02 - CONEXÃO COM BANCO DE DADOS

uv add sqlalchemy psycopg2 --default-index https://pypi.org/simple/

Install e Import bibliotecas `dotenv e os` para armazenar dados sensíveis, ex.: senhas

uv add python-dotenv

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

Criar banco de dados no PostgreSQL (pgAdmin 4): <br>
db_cash_flow_uv

```python
db_password = os.getenv("DB_PASSWORD")
```

```python
from sqlalchemy import create_engine

# Conexão
usuario = "postgres"
senha = db_password
host = "localhost"
porta = "5432"
banco = "db_cash_flow_uv"

# Criando engine de conexão
engine = create_engine(f"postgresql+psycopg2://{usuario}:{senha}@{host}:{porta}/{banco}")
```

Etapa para enviar os DataFrames para o banco de dados "db_cash_flow_uv" no PostgreSQL (pgAdmin 4).

```python
nome_tabela_1 = "bancos"
nome_tabela_2 = "plano_contas"
nome_tabela_3 = "movimentos"
nome_tabela_4 = "saldo_anterior"

df_bancos.to_sql(nome_tabela_1, engine, if_exists='replace', index=False)
df_plano_contas.to_sql(nome_tabela_2, engine, if_exists='replace', index=False)
df_movimentos.to_sql(nome_tabela_3, engine, if_exists='replace', index=False)
df_saldo_anterior.to_sql(nome_tabela_4, engine, if_exists='replace', index=False)
```
